# 调用线上的大模型
## 1. 调用具体的厂商的大模型
### 1.1 调用deepseek模型

调用大模型 涉及到三个参数信息
base_url =  model_name  =  api_key


举例1= 将参数信息存储在配置文件 => .env




```python
import os

from dotenv import load_dotenv
from langchain_deepseek import ChatDeepSeek

# 1 读取 .env 配置文件中的信息
load_dotenv(override=True) # 相关的环境变量 以.env文件中的优先


DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 2 模型的初始化
llm_deepseek = ChatDeepSeek(
    model="deepseek-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
)

# 3 模型的调用
response= llm_deepseek.invoke("请用一句话介绍你自己")
print(response)


```

## 2. 兼容的写法

使用ChatOpenAI

举例1：调用deepseek官网提供的大模型

```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

# 加载配置文件
load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 获取大模型
model = ChatOpenAI(
    model="deepseek-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
)

print(model.invoke("1 + 1 = ？"))
```

## 4. 使用LangChain1.x的统一方式

使用 init_chat_model()

举例1：调用DeepSeek官网的DeepSeek模型

```python
from langchain.chat_models import init_chat_model
import os
from dotenv import load_dotenv

# 加载配置文件
load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 获取大模型
model = init_chat_model(
    # model="deepseek-v4-flash",
    # model_provider="deepseek",
    # 或者
    model="deepseek:deepseek-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
)

print(model.invoke("一句话介绍下你自己"))
```
